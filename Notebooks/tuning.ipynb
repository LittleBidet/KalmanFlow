```python
from pathlib import Path

from reservoir_pandas_api.prepare_reservoir_data import (
    prepare_reservoir_data,
    sources_for_reservoir,
)
from kalmone import tune_inflow_model

project_root = Path.cwd().parent
sources = sources_for_reservoir(project_root, "Lexington")

prepared = prepare_reservoir_data(
    sources,
    start="2023-01-01T00:00:00Z",
    end="2026-01-01T00:00:00Z",
)

result = tune_inflow_model(
    prepared.observations,
    reservoir_id="lexington",
    reservoir_name="Lexington Reservoir",
    max_evaluations=64,
    max_tuning_rows=5_000,
    random_state=42,
)

print(result.parameters)
print(result.score)

result.config
result.model_output
```

```python
from kalmone import save_reservoir_configs

save_reservoir_configs(
    {"lexington": result.config},
    "reservoir-configurations.json",
)
```
